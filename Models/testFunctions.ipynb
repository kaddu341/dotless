```python
text = "أسبوعك"
```

```python
def build_dotless_maps():
    # Standard Map for everything EXCEPT Nun
    swift_map = {
        "ا": ["ا", "أ", "إ", "آ", "ء", "اء", "آء"],
        "ٮ": ["ب", "ت", "ث"], # Note: Nun is handled separately
        "ح": ["ج", "ح", "خ"],
        "د": ["د", "ذ"],
        "ر": ["ر", "ز"],
        "س": ["س", "ش"],
        "ص": ["ص", "ض"],
        "ط": ["ط", "ظ"],
        "ع": ["ع", "غ"],
        "ڡ": ["ف", "ق"],
        "ك": ["ك"],
        "ل": ["ل"],
        "م": ["م"],
        "و": ["و", "ؤ", "وء"],
        "ه": ["ه", "ة"],
        "ى": ["ي", "ى", "ئ", "يء"]
    }
    
    to_dotless = {}
    for dotless_char, variants in swift_map.items():
        for dotted in variants:
            to_dotless[dotted] = dotless_char
            
    return to_dotless

# --- 2. CONVERSION LOGIC ---

DOTLESS_MAP = build_dotless_maps()
# Define what counts as a "separator" (breaks a word)
SEPARATORS = set(string.whitespace + string.punctuation + "،؛؟.")

def to_dotless(text):
    result = []
    length = len(text)
    
    for i, char in enumerate(text):
        # RULE: Handle Nun (ن) dynamically
        if char == 'ن':
            # Check if it's the last char OR followed by a separator
            is_end_of_word = (i == length - 1) or (text[i+1] in SEPARATORS)
            
            if is_end_of_word:
                result.append("ں") # Distinct End Shape
            else:
                result.append("ٮ") # Connects like Ba
        
        # RULE: Handle everything else via the map
        else:
            result.append(DOTLESS_MAP.get(char, char))
            
    return "".join(result)

            
converted = to_dotless(text)
print(f"Output: {converted}")
print("-" * 30)

```
