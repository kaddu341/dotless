```python
def check_for_root(tokens):
    tokens = [w.replace('#', '') for w in tokens]
    roots = []
    for w in tokens:
        stem = ArListem.light_stem(w)
        root = ArListem.get_root()
        roots.append(root)

    is_roots = [w in rootlist for w in roots]
    any_true_root = any(is_roots)
    all_true_root = all(is_roots)
    real_roots = [roots[i] for i in range(len(roots)) if is_roots[i]]
    return any_true_root, all_true_root, is_roots, roots, real_roots
```

```python
import torch
import torch.nn.functional as F

def get_candidate_word_probabilities(input_text, candidate_words):
    # Ensure the model is in evaluation mode (disables dropout, etc.)
    model.eval()

    tokenized_text = tokenizer.tokenize(input_text)
    
    try:
        masked_word_index = tokenized_text.index('[MASK]')
    except ValueError:
        print("Error: Input text must contain '[MASK]' token.")
        return {}

    input_ids = tokenizer.convert_tokens_to_ids(tokenized_text)
    
    # Convert to PyTorch tensor and add batch dimension (1, seq_len)
    input_ids_tensor = torch.tensor([input_ids])
    
    # Perform prediction
    with torch.no_grad():
        output = model(input_ids_tensor)
    
    # Extract logits for the masked word
    # output.logits shape: (batch_size, seq_len, vocab_size)
    mask_token_logits = output.logits[0, masked_word_index, :]
    
    # Apply softmax to get probabilities across the vocabulary
    predictions = F.softmax(mask_token_logits, dim=0)

    # Tokenize and verify candidate words
    pre_tokenized_candidate_words = [tokenizer.tokenize(word) for word in candidate_words]
    tokenized_candidate_words = []
    verified_words = []

    for idx, tokens in enumerate(pre_tokenized_candidate_words):
        rep_word = candidate_words[idx]
        
        # Case 1: Word is a single token
        if len(tokens) == 1:
            tokenized_candidate_words.append(tokens[0])
            verified_words.append(rep_word)
        else:
            # Case 2: Multi-token word, try stemming (Assuming ArListem is imported)
            # if rep_word not in arDictionary: continue
            try:
                stem = ArListem.light_stem(rep_word)
                root = ArListem.get_root()
                
                stem_tokens = tokenizer.tokenize(stem)
                root_tokens = tokenizer.tokenize(root)
                
                if len(stem_tokens) == 1:
                    tokenized_candidate_words.append(stem_tokens[0])
                    verified_words.append(rep_word)
                elif len(root_tokens) == 1:
                    tokenized_candidate_words.append(root_tokens[0])
                    verified_words.append(rep_word)
            except NameError:
                print("Warning: ArListem is not defined.")
                continue

    # Convert tokens to IDs
    # convert_tokens_to_ids returns an int for a single string in most HF tokenizers
    candidate_word_ids = [tokenizer.convert_tokens_to_ids(word) for word in tokenized_candidate_words]

    # Calculate probabilities for each candidate word
    # .item() converts the 0-dim tensor to a standard Python float
    candidate_probabilities = {
        word: predictions[word_id].item() 
        for word, word_id in zip(verified_words, candidate_word_ids)
    }

    return candidate_probabilities
```

```python


def generate_probabilties(example, gen_prob_func=get_candidate_word_probabilities):
    input_text = example["Masked"]
    candidates = example["Options"]

    word_probabilities = gen_prob_func(input_text, candidates)

    sorted_words = sorted(word_probabilities, key=word_probabilities.get, reverse=True)
    if len(sorted_words) > 0:
        most_probable_word = sorted_words[0]
    else:
        most_probable_word = None
        #print(example["Target"])

    return word_probabilities, sorted_words, most_probable_word
```
